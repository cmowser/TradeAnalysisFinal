```python
import pandas as pd
import duckdb
```

```python
# Load the historical player box scores.
player_Stats = duckdb.read_parquet("../data/raw/PlayerStats.parquet").df()
```

```python
player_Stats.head()
```

```python
# Separate the source timestamp into date and time fields.
player_Stats[["game_Date", "game_Time"]] = player_Stats["gameDate"].str.split(" ", expand=True)
```

```python
# Keep regular-season player games in the transaction window.
query = """
    SELECT * FROM player_Stats
    WHERE game_Date BETWEEN '1976-10-21' AND '2019-04-10'
    AND gameType = 'Regular Season'
    """
```

```python
select_Games = duckdb.sql(query).df().reset_index()
```

```python
select_Games.head()
```

```python
select_Games.tail()
```

```python
# Save the filtered player box scores.
select_Games.to_parquet("../data/interim/filtered_Player_Stats.parquet")
```

```python
# Load the historical team box scores.
team_Stats = pd.read_csv("../data/raw/TeamStatistics.csv")
```

```python
# Separate team-game timestamps into date and time fields.
team_Stats[["game_Date", "game_Time"]] = team_Stats["gameDate"].str.split(" ", expand=True)
```

```python
# Keep regular-season team games in the transaction window.
query2 = """
    SELECT * FROM team_Stats
    WHERE game_Date BETWEEN '1976-10-21' AND '2019-04-10'
    AND gameType = 'Regular Season'
    """
```

```python
select_Team_Games = duckdb.sql(query2).df().reset_index()
```

```python
select_Team_Games.head()
```

```python
select_Team_Games.tail()
```

```python
# Save the filtered team context.
select_Team_Games.to_parquet("../data/interim/filtered_Team_Stats.parquet")
```
